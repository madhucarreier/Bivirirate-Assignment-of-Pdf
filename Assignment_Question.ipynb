```python
#9)The code takes your large salary numbers and rescales them into smaller standard numbers (called Z-scores) so that the average becomes 0 and the spread becomes 1,
#then draws a bell-curve graph.
import seaborn as sns
import matplotlib.pyplot as plt

def stdNBgraph(dataset):
    # 1. Calculate Mean and Standard Deviation of the salary column
    mean = dataset["salary"].mean()
    std = dataset["salary"].std()

    #Convert each salary value to a Z-score: (x - mean) / std
    z_score = [(x - mean) / std for x in dataset["salary"]]

    # Plot the Standard Normal Distribution
    sns.histplot(z_score, kde=True, color='blue')
    plt.title("Standard Normal Distribution of Salary (Z-scores)")
    plt.xlabel("Z-score (Standard Deviations from Mean)")
    plt.ylabel("Frequency")
    plt.show()

    return z_score

# Call the function and save the standardized values
dataset["salary_zscore"] = stdNBgraph(dataset)

# Verify that Mean is ~0 and Standard Deviation is ~1
print("New Mean:", round(dataset["salary_zscore"].mean(), 2))
print("New Standard Deviation:", round(dataset["salary_zscore"].std(), 2))
```

```python
#10) What is the probability Density Function of the salary range from 700000 to 900000?
#draws a graph of all student salaries, marks the ₹7,00,000 to ₹9,00,000 range with two red lines
#calculates the exact probability (chance) of a student getting a salary inside that range.
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import norm

def get_pdf_probability(dataset, startrange, endrange):
    # 1. Plot the distribution and vertical boundary lines
    ax = sns.histplot(dataset, kde=True, color='Green', stat='density')
    plt.axvline(startrange, color='Red', linestyle='--')
    plt.axvline(endrange, color='Red', linestyle='--')
    plt.title(f"Salary PDF between {startrange} and {endrange}")
    plt.show()
    
    # 2. Calculate sample Mean and Standard Deviation
    sample_mean = dataset.mean()
    sample_std = dataset.std()
    print(f"Mean = {sample_mean:.3f}, Standard Deviation = {sample_std:.3f}")
    
    # 3. Define the normal distribution
    dist = norm(sample_mean, sample_std)
    
    # 4. Calculate PDF values for each integer in the range and sum them
    values = [value for value in range(startrange, endrange)]
    probabilities = [dist.pdf(value) for value in values]
    prob = sum(probabilities)
    
    print(f"The area between range ({startrange}, {endrange}): {prob}")
    return prob

# Run the function for the salary range 700,000 to 900,000
get_pdf_probability(dataset["salary"], 700000, 900000)
```

```python
#11)Test the similarity between the degree_t(Sci&Tech)with respect to
#etest_p and mba_p at significance level of 5%.(Make decision using Hy
#pothesis Testing)from scipy.stats import ttest_rel

#code picks only the Science & Tech students, compares their Employability Test scores (etest_p) with their MBA scores (mba_p)
#checks if both scores the same or genuinely different.

from scipy.stats import ttest_rel

# Code picks only the Science & Tech students, compares their Employability Test scores (etest_p) with their MBA scores (mba_p)
# Checks if both scores are the same or genuinely different.

# 1. Filter only 'Sci&Tech' degree students and select their etest_p and mba_p scores
sci_tech_etest = dataset[dataset['degree_t'] == 'Sci&Tech']['etest_p']
sci_tech_mba = dataset[dataset['degree_t'] == 'Sci&Tech']['mba_p']

# 2. Perform Paired T-Test (ttest_rel)
t_stat, p_val = ttest_rel(sci_tech_etest, sci_tech_mba)

print("T-statistic:", t_stat)
print("p-value:", p_val)

# 3. Decision Rule at 5% significance level (alpha = 0.05)
if p_val < 0.05:
    print("Reject Null Hypothesis (H0). There is NO similarity (significant difference).")
else:
    print("reject ALternative Hypothesis (H0). There IS similarity between etest_p and mba_p.")
```
